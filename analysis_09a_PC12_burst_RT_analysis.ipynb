```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import os.path as op
import pickle
import json
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import RobustScaler
import pandas as pd
from scipy.ndimage import gaussian_filter1d
import mne
from scipy import stats
from functools import partial

data_dir = Path("/home/qmoreau/gogo_bursts/stats_new")
pc = 12
epoch = "STIM"

short_color=np.array([255,0,0])/255.0
long_color=np.array([0,0,255])/255.0

CONDITION_COLORS = {"SHORT": short_color, "LONG": long_color}
XLIM  = (-0.5, 1.0)
ALPHA = 0.15

json_file = "settings.json"
with open(json_file) as pipeline_file:
    parameters = json.load(pipeline_file)
path = parameters["dataset_path"]
epoch_types = ['STIM', 'RESP']
condition_names = ['SHORT', 'LONG']

all_bursts = {
    'subject_id': [],
    'group': [],
    'epoch_type': [],
    'condition': [],
    'cluster': [],
    'channel': [],
    'trial': [],
    'waveform': np.zeros((0, 156)),
    'waveform_times': [],
    'peak_freq': [],
    'peak_amp_iter': [],
    'peak_amp_base': [],
    'peak_time': [],
    'peak_adjustment': [],
    'fwhm_freq': [],
    'fwhm_time': [],
    'polarity': [],
}

df = pd.read_csv(op.join(path, 'data_v2', 'GOGO_Demographics_2025_COMO.csv'))
for subject_id, group in df.loc[:, ["ParticipantID", "Status"]].itertuples(index=False, name=None):
    subject_data_dir = op.join('/home/common/bonaiuto/gogo_bursts/derivatives_v2/processed', subject_id)
    fname = op.join(subject_data_dir, f'{subject_id}_bursts.pickle')
    if op.exists(fname):
        with open(fname, 'rb') as file:
            bursts = pickle.load(file)
        for key in bursts.keys():
            if key == 'waveform_times':
                all_bursts[key] = bursts[key]
            elif key == 'waveform':
                all_bursts[key] = np.vstack([all_bursts[key], bursts[key]])
            else:
                all_bursts[key] = np.hstack([all_bursts[key], bursts[key]])

df = pd.read_csv(op.join(path, 'data_v2', 'GOGO_Demographics_2025_Driving.csv'))
for subject_id, group in df.loc[:, ["ParticipantID", "Status"]].itertuples(index=False, name=None):
    subject_data_dir = op.join('/home/common/bonaiuto/gogo_bursts/derivatives_v2/processed', subject_id)
    fname = op.join(subject_data_dir, f'{subject_id}_bursts.pickle')
    if op.exists(fname):
        with open(fname, 'rb') as file:
            bursts = pickle.load(file)
        for key in bursts.keys():
            if key == 'waveform_times':
                all_bursts[key] = bursts[key]
            elif key == 'waveform':
                all_bursts[key] = np.vstack([all_bursts[key], bursts[key]])
            else:
                all_bursts[key] = np.hstack([all_bursts[key], bursts[key]])

burst_idx = (all_bursts['group'] == 'TD') & (all_bursts['cluster'] == 'contra')
for key in all_bursts.keys():
    if key == 'waveform_times':
        continue
    elif key == 'waveform':
        all_bursts[key] = all_bursts[key][burst_idx, :]
    else:
        all_bursts[key] = all_bursts[key][burst_idx]

waveform_times = all_bursts['waveform_times']
subject_ids = np.unique(all_bursts['subject_id'])
print(f'N subjects={len(subject_ids)}')

# --- PCA on burst waveforms (no significance-testing step, PC12 only needed) ---
waveforms = all_bursts['waveform']
burst_std = np.zeros(waveforms.shape)
for subject in np.unique(all_bursts['subject_id']):
    s_idx = all_bursts['subject_id'] == subject
    burst_std[s_idx, :] = RobustScaler().fit_transform(waveforms[s_idx, :])
burst_std = RobustScaler().fit_transform(burst_std)

model = PCA(n_components=20)
model.fit(burst_std)
pc_scores = model.transform(burst_std)
pc_labels = ["PC_{}".format(i + 1) for i in range(pc_scores.shape[1])]


n_q = 3
quartiles = np.linspace(0, 100, num=n_q + 1)
quartiles = list(zip(quartiles[:-1], quartiles[1:]))
col_range = [plt.get_cmap('winter')(x) for x in np.linspace(0, 1, n_q)]
mean_waveform = np.mean(waveforms, axis=0)

# --- sliding-window GLMM trend output (Option A: no cluster permutation) ------
trends = (
    pd.read_csv(data_dir / f"PC_{pc}_{epoch}_RT_trends_tw.csv")
    .assign(tertile=lambda d: "Tertile " + d["tertile"].astype(str))
)
trends = trends[trends["SE"].apply(lambda x: pd.notna(x) and np.isfinite(x) and x != 0)]

contrasts = (
    pd.read_csv(data_dir / f"PC_{pc}_{epoch}_RT_contrasts_tw.csv")
    .assign(tertile=lambda d: "Tertile " + d["tertile"].astype(str))
)
contrasts = contrasts[contrasts["SE"].apply(lambda x: pd.notna(x) and np.isfinite(x) and x != 0)]

# rebuild a per-window significance table (uncorrected p < .05, one row per
# tertile x time) from the trends (per-condition slope + p) and contrasts
# (LONG-SHORT + p) files — replaces the old cluster-permutation windows CSV
sig_long  = trends[trends["condition"] == "LONG"][["tertile", "time", "p.value"]].rename(columns={"p.value": "long_p"})
sig_short = trends[trends["condition"] == "SHORT"][["tertile", "time", "p.value"]].rename(columns={"p.value": "short_p"})
sig_ctr   = contrasts[["tertile", "time", "p.value"]].rename(columns={"p.value": "long_short_p"})

windows = (
    sig_long.merge(sig_short, on=["tertile", "time"], how="outer")
            .merge(sig_ctr, on=["tertile", "time"], how="outer")
            .sort_values(["tertile", "time"])
)

# mean RT per condition (across all tertiles and subjects)
trials = pd.read_csv(data_dir / f"PC_{pc}_{epoch}_trial_burst_counts.csv")
mean_rt = (
    trials.dropna(subset=["response_time"])
    .groupby("condition")["response_time"]
    .mean()
)
print("Mean RT per condition (s):")
print(mean_rt)

tertiles = sorted(trends["tertile"].unique())

trends_vis = trends[trends["time"].between(*XLIM)]
y_min  = trends_vis["estimate"].min()
y_max  = trends_vis["estimate"].max()
y_span = y_max - y_min
y_pad  = 0.2 * y_span

py_long  = y_max + 1.0 * y_pad
py_short = y_max + 1.8 * y_pad
py_ctr   = y_max + 2.6 * y_pad

fig, axes = plt.subplots(
    1, len(tertiles),
    figsize=(12, 4.5),
    sharey=True,
    constrained_layout=True,
)
if len(tertiles) == 1:
    axes = [axes]

scores = pc_scores[:, pc-1] # O indexed here

for tert_ix, (ax, tert) in enumerate(zip(axes, tertiles)):
    t_df = trends[trends["tertile"] == tert]
    w_df = windows[windows["tertile"] == tert]
    w_df = w_df[(w_df["time"] >= XLIM[0]) & (w_df["time"] <= XLIM[1])]

    # === PRINT STATISTICS FOR THIS BURST TYPE (uncorrected p < .05) ===
    print(f"\n{'='*70}")
    print(f"BURST TYPE {tert_ix+1} (Tertile {tert})")
    print(f"{'='*70}\n")

    sig_long_t  = w_df.loc[w_df["long_p"] < 0.05, "time"].tolist()
    sig_short_t = w_df.loc[w_df["short_p"] < 0.05, "time"].tolist()
    sig_ctr_t   = w_df.loc[w_df["long_short_p"] < 0.05, "time"].tolist()

    if not sig_long_t and not sig_short_t and not sig_ctr_t:
        print("No significant time points found (uncorrected p < .05).\n")
    else:
        print("Significant time points (uncorrected p < .05):\n")
        if sig_long_t:
            print(f"  LONG:       {[f'{t:.3f}' for t in sig_long_t]}")
        if sig_short_t:
            print(f"  SHORT:      {[f'{t:.3f}' for t in sig_short_t]}")
        if sig_ctr_t:
            print(f"  SHORT-LONG: {[f'{t:.3f}' for t in sig_ctr_t]}")
        print()

    ax.axhline(0, color="k", linestyle="--", linewidth=0.8, zorder=1)
    ax.axvline(0, color="k", linestyle="--", linewidth=0.8, zorder=1)

    # mean RT vertical lines per condition
    for cond, color in CONDITION_COLORS.items():
        if cond in mean_rt.index:
            ax.vlines(mean_rt[cond], ymin=y_min - y_pad, ymax=0.01,
                      color=color, linestyle=":", linewidth=1.5, zorder=2)

    # ribbon + line per condition
    for cond, color in CONDITION_COLORS.items():
        cdf = t_df[t_df["condition"] == cond].sort_values("time")
        ax.fill_between(
            cdf["time"],
            cdf["estimate"] - cdf["SE"],
            cdf["estimate"] + cdf["SE"],
            color=color, alpha=ALPHA, linewidth=0,
        )
        ax.plot(cdf["time"], cdf["estimate"],
                color=color, linewidth=1.5, label=cond)

    ax.set_ylim(y_min - y_pad, y_max + 4 * y_pad)

    # stars at each significant timepoint (uncorrected, no cluster ranges)
    for _, win in w_df.iterrows():
        t = win["time"]
        if pd.notna(win["long_p"]) and win["long_p"] < 0.05:
            ax.text(t, py_long,  "*", color=long_color,
                    ha="center", va="bottom", fontsize=8)
        if pd.notna(win["short_p"]) and win["short_p"] < 0.05:
            ax.text(t, py_short, "*", color=short_color,
                    ha="center", va="bottom", fontsize=8)
        if pd.notna(win["long_short_p"]) and win["long_short_p"] < 0.05:
            ax.text(t, py_ctr,   "*", color="purple",
                    ha="center", va="bottom", fontsize=8)

    ax.set_xlim(XLIM)
    ax.set_title(f"Burst type {tert_ix+1}", fontweight="bold", fontsize=10, pad=28)
    ax.set_xlabel("Time (s)", fontsize=9)
    ax.xaxis.set_major_locator(plt.MaxNLocator(6))
    ax.spines[["top", "right"]].set_visible(False)

    # --- waveform strip above the plot, between title and axes ---
    b, e = quartiles[tert_ix]
    q_map = (scores > np.percentile(scores, b)) & (scores <= np.percentile(scores, e))
    q_mean = np.mean(waveforms[q_map], axis=0)

    inset = ax.inset_axes([0.2, 1.02, 0.6, 0.18], transform=ax.transAxes)
    inset.plot(waveform_times, q_mean / 1e-15, color=col_range[tert_ix], linewidth=1.2)
    inset.set_xticks([])
    inset.set_yticks([])
    inset.set_facecolor("none")
    for spine in inset.spines.values():
        spine.set_visible(False)

axes[0].set_ylabel("Burst count slope on RT", fontsize=9)

# legend: conditions + RT lines
handles, labels = axes[0].get_legend_handles_labels()
for cond, color in CONDITION_COLORS.items():
    handles.append(plt.Line2D([0], [0], color=color, linestyle=":", linewidth=1.5))
    labels.append(f"Mean RT {cond}")

fig.legend(handles, labels,
           loc="upper center", ncol=len(CONDITION_COLORS) * 2,
           frameon=False, fontsize=9,
           bbox_to_anchor=(0.5, 1.1))

fig.suptitle(f"PC{pc} - burst type vs RT ({epoch})", fontsize=11, y=1.16)

out_path = data_dir / f"PC{pc}_{epoch}_burst_RT_windows.pdf"
fig.savefig(out_path, bbox_inches="tight", dpi=150)
print(f"Saved - {out_path}")
plt.show()

import mne
from scipy.ndimage import gaussian_filter1d, label as ndlabel

condition_names = ['SHORT', 'LONG']
epoch_types     = ['STIM', 'RESP']

stim_idx = all_bursts['epoch_type'] == 'STIM'
resp_idx = all_bursts['epoch_type'] == 'RESP'

bin_width      = 0.025
stim_time_bins = np.arange(-1.75, 1.75, bin_width)
resp_time_bins = np.arange(-1.75, 1.75, bin_width)

def print_cluster_results(T_obs, clusters, cluster_p_values, freqs, times, label):
    print(f"\n{'='*50}")
    print(f"Cluster results: {label}")
    print(f"{'='*50}")

    sig_mask = np.zeros(T_obs.shape, dtype=bool)
    for clu, p in zip(clusters, cluster_p_values):
        if p < 0.05:
            sig_mask[clu] = True

    if not sig_mask.any():
        print("No significant clusters (p < .05)")
        return

    labeled, n_regions = ndlabel(sig_mask)
    print(f"  {n_regions} contiguous significant region(s) found\n")

    for region_idx in range(1, n_regions + 1):
        region_mask = labeled == region_idx
        indices = np.where(region_mask)
        t_vals = T_obs[region_mask]
        region_ps = [p for clu, p in zip(clusters, cluster_p_values)
                     if p < 0.05 and region_mask[clu].any()]

        print(f"  Region #{region_idx}:")
        print(f"    p range: [{min(region_ps):.4f}, {max(region_ps):.4f}]")
        print(f"    T range: [{t_vals.min():.3f}, {t_vals.max():.3f}]")

        if T_obs.ndim == 2:
            freq_idx, time_idx = indices
            print(f"    Frequency: {freqs[freq_idx.min()]:.1f} - {freqs[freq_idx.max()]:.1f} Hz")
            print(f"    Time: {times[time_idx.min()]:.3f} - {times[time_idx.max()]:.3f} s")
        else:
            time_idx = indices[0]
            print(f"    Time: {times[time_idx.min()]:.3f} - {times[time_idx.max()]:.3f} s")

        print(f"    Size (pixels): {region_mask.sum()}")

# -- PC12-tertile burst rate (smoothed, as in the reference) ---------------
step = 100 / n_q
stim_plt_idx = (stim_time_bins[:-1] >= -1.5) & (stim_time_bins[:-1] <= 1.5)
resp_plt_idx = (resp_time_bins[:-1] >= -.5)  & (resp_time_bins[:-1] <= 1.5)
stim_time_plt = stim_time_bins[:-1][stim_plt_idx]
resp_time_plt = resp_time_bins[:-1][resp_plt_idx]
base_idx = np.where((stim_time_bins >= -1.5) & (stim_time_bins < -1))[0]

threshold_tfce = dict(start=0, step=0.005)

cmap = plt.get_cmap('winter')
col_range = [cmap(x) for x in np.linspace(0, 1, n_q)]

q_bins = np.percentile(pc_scores[:, pc-1], np.arange(0, 100 + step, step))

q_rates_smoothed = {
    q: {ep: {c: [] for c in condition_names} for ep in epoch_types}
    for q in range(n_q)
}

for subject_id in subject_ids:
    subj_idx = (all_bursts['subject_id'] == subject_id)
    stim_b_idx = subj_idx & stim_idx
    resp_b_idx = subj_idx & resp_idx

    for q in range(n_q):
        q_idx = ((pc_scores[:, pc-1] >= q_bins[q]) & (pc_scores[:, pc-1] <= q_bins[q + 1]))

        stim_short_idx = stim_b_idx & (all_bursts['condition'] == 'SHORT')
        resp_short_idx = resp_b_idx & (all_bursts['condition'] == 'SHORT')
        stim_long_idx  = stim_b_idx & (all_bursts['condition'] == 'LONG')
        resp_long_idx  = resp_b_idx & (all_bursts['condition'] == 'LONG')

        stim_short_trials = np.unique(all_bursts['trial'][stim_short_idx])
        resp_short_trials = np.unique(all_bursts['trial'][resp_short_idx])
        stim_long_trials  = np.unique(all_bursts['trial'][stim_long_idx])
        resp_long_trials  = np.unique(all_bursts['trial'][resp_long_idx])

        stim_short_rate, _ = np.histogram(
            all_bursts['peak_time'][subj_idx & stim_short_idx & q_idx], bins=stim_time_bins)
        resp_short_rate, _ = np.histogram(
            all_bursts['peak_time'][subj_idx & resp_short_idx & q_idx], bins=resp_time_bins)
        stim_long_rate, _ = np.histogram(
            all_bursts['peak_time'][subj_idx & stim_long_idx & q_idx], bins=stim_time_bins)
        resp_long_rate, _ = np.histogram(
            all_bursts['peak_time'][subj_idx & resp_long_idx & q_idx], bins=resp_time_bins)

        stim_short_rate = (stim_short_rate / bin_width) / len(stim_short_trials)
        resp_short_rate = (resp_short_rate / bin_width) / len(resp_short_trials)
        stim_long_rate  = (stim_long_rate / bin_width) / len(stim_long_trials)
        resp_long_rate  = (resp_long_rate / bin_width) / len(resp_long_trials)

        stim_short_rate = gaussian_filter1d(stim_short_rate, sigma=2.5, axis=0, mode="nearest")
        resp_short_rate = gaussian_filter1d(resp_short_rate, sigma=2.5, axis=0, mode="nearest")
        stim_long_rate  = gaussian_filter1d(stim_long_rate, sigma=2.5, axis=0, mode="nearest")
        resp_long_rate  = gaussian_filter1d(resp_long_rate, sigma=2.5, axis=0, mode="nearest")

        base_rate = np.mean(stim_short_rate[base_idx])
        stim_short_rate = (stim_short_rate - base_rate)[stim_plt_idx]
        resp_short_rate = (resp_short_rate - base_rate)[resp_plt_idx]
        base_rate = np.mean(stim_long_rate[base_idx])
        stim_long_rate = (stim_long_rate - base_rate)[stim_plt_idx]
        resp_long_rate = (resp_long_rate - base_rate)[resp_plt_idx]

        q_rates_smoothed[q]['STIM']['SHORT'].append(stim_short_rate)
        q_rates_smoothed[q]['RESP']['SHORT'].append(resp_short_rate)
        q_rates_smoothed[q]['STIM']['LONG'].append(stim_long_rate)
        q_rates_smoothed[q]['RESP']['LONG'].append(resp_long_rate)

# -- Figure 1: waveform, standalone -----------------------------------------
fig_wave, ax_wave = plt.subplots(figsize=(5, 4))
ax_wave.set_title(f'PC{pc}')
scores = pc_scores[:, pc-1]
for q_ix, (b, e) in enumerate(quartiles):
    q_map = (scores > np.percentile(scores, b)) & (scores <= np.percentile(scores, e))
    q_mean = np.mean(waveforms[q_map], axis=0)
    ax_wave.plot(waveform_times, q_mean / 1e-15, lw=2, c=col_range[q_ix], label=f'Burst type {q_ix+1}')
ax_wave.set_xlabel('Time (s)')
ax_wave.set_ylabel('Amplitude (fT)')
ax_wave.legend()
plt.tight_layout()
plt.show()

# -- Figure 2: burst rates, 2x2 grid, with cluster stats --------------------
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 7))

# STIM SHORT
for q in range(n_q):
    m_rate = np.mean(q_rates_smoothed[q]['STIM']['SHORT'], axis=0)
    se_rate = np.std(q_rates_smoothed[q]['STIM']['SHORT'], axis=0) / np.sqrt(len(q_rates_smoothed[q]['STIM']['SHORT']))
    ax1.plot(stim_time_plt, m_rate, color=col_range[q])
    ax1.fill_between(stim_time_plt, m_rate - se_rate, m_rate + se_rate, alpha=0.25, color=col_range[q])

for q in range(n_q):
    stim_data = np.array(q_rates_smoothed[q]['STIM']['SHORT'])
    stim_T_obs, stim_clusters, stim_cluster_p_values, H0 = mne.stats.permutation_cluster_1samp_test(
        stim_data, threshold=threshold_tfce, n_permutations=1000, tail=0, n_jobs=-1
    )
    print_cluster_results(stim_T_obs, stim_clusters, stim_cluster_p_values, None, stim_time_plt,
                          f"PC{pc} Burst type {q+1} - STIM SHORT")
    yl = ax1.get_ylim()
    py = yl[1] + (q + 1) * .05 * (yl[1] - 0)
    for p_idx, p_val in enumerate(stim_cluster_p_values):
        if p_val < 0.05:
            ax1.plot(stim_time_plt[p_idx], py, '*', color=col_range[q])
ax1.set_title('STIM - SHORT')
ax1.axvline(0, color='k', linestyle='--')
ax1.set_xlabel('Time (s)')
ax1.set_ylabel(r'$\Delta$ burst rate (Hz)')
yl1 = ax1.get_ylim()

# RESP SHORT
for q in range(n_q):
    m_rate = np.mean(q_rates_smoothed[q]['RESP']['SHORT'], axis=0)
    se_rate = np.std(q_rates_smoothed[q]['RESP']['SHORT'], axis=0) / np.sqrt(len(q_rates_smoothed[q]['RESP']['SHORT']))
    ax2.plot(resp_time_plt, m_rate, color=col_range[q], label=f'Burst type {q+1}')
    ax2.fill_between(resp_time_plt, m_rate - se_rate, m_rate + se_rate, alpha=0.25, color=col_range[q])

for q in range(n_q):
    resp_data = np.array(q_rates_smoothed[q]['RESP']['SHORT'])
    resp_T_obs, resp_clusters, resp_cluster_p_values, H0 = mne.stats.permutation_cluster_1samp_test(
        resp_data, threshold=threshold_tfce, n_permutations=1000, tail=0, n_jobs=-1
    )
    print_cluster_results(resp_T_obs, resp_clusters, resp_cluster_p_values, None, resp_time_plt,
                          f"PC{pc} Burst type {q+1} - RESP SHORT")
    yl = ax2.get_ylim()
    py = yl[1] + (q + 1) * .05 * (yl[1] - 0)
    for p_idx, p_val in enumerate(resp_cluster_p_values):
        if p_val < 0.05:
            ax2.plot(resp_time_plt[p_idx], py, '*', color=col_range[q])
ax2.legend()
ax2.set_title('RESP - SHORT')
ax2.axvline(0, color='k', linestyle='--')
ax2.set_xlabel('Time (s)')
ax2.set_ylabel(r'$\Delta$ burst rate (Hz)')
yl2 = ax2.get_ylim()

# STIM LONG
for q in range(n_q):
    m_rate = np.mean(q_rates_smoothed[q]['STIM']['LONG'], axis=0)
    se_rate = np.std(q_rates_smoothed[q]['STIM']['LONG'], axis=0) / np.sqrt(len(q_rates_smoothed[q]['STIM']['LONG']))
    ax3.plot(stim_time_plt, m_rate, color=col_range[q])
    ax3.fill_between(stim_time_plt, m_rate - se_rate, m_rate + se_rate, alpha=0.25, color=col_range[q])

for q in range(n_q):
    stim_data = np.array(q_rates_smoothed[q]['STIM']['LONG'])
    stim_T_obs, stim_clusters, stim_cluster_p_values, H0 = mne.stats.permutation_cluster_1samp_test(
        stim_data, threshold=threshold_tfce, n_permutations=1000, tail=0, n_jobs=-1
    )
    print_cluster_results(stim_T_obs, stim_clusters, stim_cluster_p_values, None, stim_time_plt,
                          f"PC{pc} Burst type {q+1} - STIM LONG")
    yl = ax3.get_ylim()
    py = yl[1] + (q + 1) * .05 * (yl[1] - 0)
    for p_idx, p_val in enumerate(stim_cluster_p_values):
        if p_val < 0.05:
            ax3.plot(stim_time_plt[p_idx], py, '*', color=col_range[q])
ax3.set_title('STIM - LONG')
ax3.axvline(0, color='k', linestyle='--')
ax3.set_xlabel('Time (s)')
ax3.set_ylabel(r'$\Delta$ burst rate (Hz)')
yl3 = ax3.get_ylim()

# RESP LONG
for q in range(n_q):
    m_rate = np.mean(q_rates_smoothed[q]['RESP']['LONG'], axis=0)
    se_rate = np.std(q_rates_smoothed[q]['RESP']['LONG'], axis=0) / np.sqrt(len(q_rates_smoothed[q]['RESP']['LONG']))
    ax4.plot(resp_time_plt, m_rate, color=col_range[q], label=f'Burst type {q+1}')
    ax4.fill_between(resp_time_plt, m_rate - se_rate, m_rate + se_rate, alpha=0.25, color=col_range[q])

for q in range(n_q):
    resp_data = np.array(q_rates_smoothed[q]['RESP']['LONG'])
    resp_T_obs, resp_clusters, resp_cluster_p_values, H0 = mne.stats.permutation_cluster_1samp_test(
        resp_data, threshold=threshold_tfce, n_permutations=1000, tail=0, n_jobs=-1
    )
    print_cluster_results(resp_T_obs, resp_clusters, resp_cluster_p_values, None, resp_time_plt,
                          f"PC{pc} Burst type {q+1} - RESP LONG")
    yl = ax4.get_ylim()
    py = yl[1] + (q + 1) * .05 * (yl[1] - 0)
    for p_idx, p_val in enumerate(resp_cluster_p_values):
        if p_val < 0.05:
            ax4.plot(resp_time_plt[p_idx], py, '*', color=col_range[q])
ax4.legend()
ax4.set_title('RESP - LONG')
ax4.axvline(0, color='k', linestyle='--')
ax4.set_xlabel('Time (s)')
ax4.set_ylabel(r'$\Delta$ burst rate (Hz)')
yl4 = ax4.get_ylim()

ymin = np.min([yl1[0], yl2[0], yl3[0], yl4[0]])
ymax = np.max([yl1[1], yl2[1], yl3[0], yl4[0]])
ax1.set_ylim(ymin, ymax)
ax2.set_ylim(ymin, ymax)
ax3.set_ylim(ymin, ymax)
ax4.set_ylim(ymin, ymax)
plt.tight_layout()
plt.show()

tertiles = sorted(trends["tertile"].unique())

trends_vis = trends[trends["time"].between(*XLIM)]
y_min  = trends_vis["estimate"].min()
y_max  = trends_vis["estimate"].max()
y_span = y_max - y_min
y_pad  = 0.2 * y_span

py_long  = y_max + 1.0 * y_pad
py_short = y_max + 1.8 * y_pad
py_ctr   = y_max + 2.6 * y_pad

fig, axes = plt.subplots(
    1, len(tertiles),
    figsize=(12, 4.5),
    sharey=True,
    constrained_layout=True,
)
if len(tertiles) == 1:
    axes = [axes]

scores = pc_scores[:, pc-1]

rate_times = stim_time_bins[:-1][stim_plt_idx] if epoch == "STIM" else resp_time_bins[:-1][resp_plt_idx]
rate_vals = [
    np.mean(q_rates_smoothed[q][epoch][cond], axis=0)
    for q in range(len(tertiles)) for cond in CONDITION_COLORS
]
rate_min, rate_max = np.min(rate_vals), np.max(rate_vals)

ax_lo, ax_hi = y_min - y_pad, y_max + 4 * y_pad
f_zero = -ax_lo / (ax_hi - ax_lo)          # fraction of the RT axis below 0

k = max(-rate_min / f_zero, rate_max / (1 - f_zero)) * 1.1   # 10% headroom
rate_ylim = (-k * f_zero, k * (1 - f_zero))

for tert_ix, (ax, tert) in enumerate(zip(axes, tertiles)):
    t_df = trends[trends["tertile"] == tert]
    w_df = windows[windows["tertile"] == tert]
    w_df = w_df[(w_df["time"] >= XLIM[0]) & (w_df["time"] <= XLIM[1])]

    ax_rate = ax.twinx()
    ax_rate.set_zorder(ax.get_zorder() - 1)
    ax.patch.set_visible(False)

    for cond, color in CONDITION_COLORS.items():
        m_rate = np.mean(q_rates_smoothed[tert_ix][epoch][cond], axis=0)
        ax_rate.plot(rate_times, m_rate, color=color, alpha=0.25, linewidth=1, zorder=0)

    ax_rate.set_ylim(rate_ylim)
    ax_rate.tick_params(axis="y", labelsize=8)
    if tert_ix == len(tertiles) - 1:
        ax_rate.set_ylabel(r"$\Delta$ burst rate (Hz)", fontsize=8, color="0.5")
        ax_rate.tick_params(axis="y", labelsize=7, colors="0.6")

    else:
        ax_rate.set_yticklabels([])

    # === PRINT STATISTICS FOR THIS BURST TYPE (uncorrected p < .05) ===
    print(f"\n{'='*70}")
    print(f"BURST TYPE {tert_ix+1} (Tertile {tert})")
    print(f"{'='*70}\n")

    sig_long_t  = w_df.loc[w_df["long_p"] < 0.05, "time"].tolist()
    sig_short_t = w_df.loc[w_df["short_p"] < 0.05, "time"].tolist()
    sig_ctr_t   = w_df.loc[w_df["long_short_p"] < 0.05, "time"].tolist()

    if not sig_long_t and not sig_short_t and not sig_ctr_t:
        print("No significant time points found (uncorrected p < .05).\n")
    else:
        print("Significant time points (uncorrected p < .05):\n")
        if sig_long_t:
            print(f"  LONG:       {[f'{t:.3f}' for t in sig_long_t]}")
        if sig_short_t:
            print(f"  SHORT:      {[f'{t:.3f}' for t in sig_short_t]}")
        if sig_ctr_t:
            print(f"  SHORT-LONG: {[f'{t:.3f}' for t in sig_ctr_t]}")
        print()

    ax.axhline(0, color="k", linestyle="--", linewidth=0.8, zorder=1)
    ax.axvline(0, color="k", linestyle="--", linewidth=0.8, zorder=1)

    # mean RT vertical lines per condition
    for cond, color in CONDITION_COLORS.items():
        if cond in mean_rt.index:
            ax.vlines(mean_rt[cond], ymin=y_min - y_pad, ymax=0.01,
                      color=color, linestyle=":", linewidth=1.5, zorder=2)

    # ribbon + line per condition
    for cond, color in CONDITION_COLORS.items():
        cdf = t_df[t_df["condition"] == cond].sort_values("time")
        ax.fill_between(
            cdf["time"],
            cdf["estimate"] - cdf["SE"],
            cdf["estimate"] + cdf["SE"],
            color=color, alpha=ALPHA, linewidth=0,
        )
        ax.plot(cdf["time"], cdf["estimate"],
                color=color, linewidth=1.5, label=cond)

    ax.set_ylim(y_min - y_pad, y_max + 4 * y_pad)

    # stars at each significant timepoint (uncorrected, no cluster ranges)
    for _, win in w_df.iterrows():
        t = win["time"]
        if pd.notna(win["long_p"]) and win["long_p"] < 0.05:
            ax.text(t, py_long,  "*", color=long_color,
                    ha="center", va="bottom", fontsize=8)
        if pd.notna(win["short_p"]) and win["short_p"] < 0.05:
            ax.text(t, py_short, "*", color=short_color,
                    ha="center", va="bottom", fontsize=8)
        if pd.notna(win["long_short_p"]) and win["long_short_p"] < 0.05:
            ax.text(t, py_ctr,   "*", color="purple",
                    ha="center", va="bottom", fontsize=8)

    ax.set_xlim(XLIM)
    ax.set_title(f"Burst type {tert_ix+1}", fontweight="bold", fontsize=10, pad=28)
    ax.set_xlabel("Time (s)", fontsize=9)
    ax.xaxis.set_major_locator(plt.MaxNLocator(6))
    ax.spines[["top", "right"]].set_visible(False)

    # --- waveform strip above the plot, between title and axes ---
    b, e = quartiles[tert_ix]
    q_map = (scores > np.percentile(scores, b)) & (scores <= np.percentile(scores, e))
    q_mean = np.mean(waveforms[q_map], axis=0)

    inset = ax.inset_axes([0.2, 1.02, 0.6, 0.18], transform=ax.transAxes)
    inset.plot(waveform_times, q_mean / 1e-15, color=col_range[tert_ix], linewidth=1.2)
    inset.set_xticks([])
    inset.set_yticks([])
    inset.set_facecolor("none")
    for spine in inset.spines.values():
        spine.set_visible(False)

axes[0].set_ylabel("Burst count slope on RT", fontsize=9)

# legend: conditions + RT lines
handles, labels = axes[0].get_legend_handles_labels()
for cond, color in CONDITION_COLORS.items():
    handles.append(plt.Line2D([0], [0], color=color, linestyle=":", linewidth=1.5))
    labels.append(f"Mean RT {cond}")

fig.legend(handles, labels,
           loc="upper center", ncol=len(CONDITION_COLORS) * 2,
           frameon=False, fontsize=9,
           bbox_to_anchor=(0.5, 1.1))

fig.suptitle(f"PC{pc} - burst type vs RT ({epoch})", fontsize=11, y=1.1)

out_path = data_dir / f"PC{pc}_{epoch}_burst_RT_overlay.pdf"
fig.savefig(out_path, bbox_inches="tight", dpi=150)
print(f"Saved - {out_path}")
plt.show()
```
